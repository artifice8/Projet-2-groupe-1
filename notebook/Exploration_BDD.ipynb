```python
import pandas as pd
lien = "https://datasets.imdbws.com/title.ratings.tsv.gz"
ratings = pd.read_csv(lien, sep="\t")
```

```python
ratings.head()
```

```python
lien2 = "https://datasets.imdbws.com/title.basics.tsv.gz"
basics = pd.read_csv(lien2, sep="\t", na_values="\\N")
```

```python
basics.head()
```

```python
test = pd.merge (basics, ratings, how='left', on='tconst')
```

```python
test = test[test['genres'].str.contains("Comedy", na=False)]
```

```python
test = test[test['titleType'].str.contains("movie", na=False)]
```
